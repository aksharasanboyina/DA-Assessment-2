```python
#function finding the area of a circle
def areaofcircle(radius):
    area=3.14*(radius**2)
    return area
print("the area of circle: ",areaofcircle(6))
```

```python
#function to find the largest number
def largestnumber(a,b,c):
    if a>b and a>c:
        return a
    elif b>a and b>c:
        return b
    else:
        return c
print("the largest number is: ",largestnumber(12,35,78))
```

```python
#function to find the length of a string using loop
def stringlen(s):
    count=0
    for i in s:
        count+=1
    return count
s=input("enter a string: ")        
print(f"the length of the string {s}: ",stringlen(s))        
```

```python
#function to check whether a number is prime or not
def isprimeornot(n):
    count=0
    for i in range(1,n+1):
        if n%i==0:
            count+=1
    if count==2:
        return("prime")
    else:
        return("not prime")
n=int(input("enter a number :"))
print(f"{n} is :",isprimeornot(n))      
```

```python
#function to find the nth fibonacci number
def Fibonacci(n):
    if n==1 or n==0:
        return n
    return Fibonacci(n-1)+Fibonacci(n-2)
num=int(input("enter the number to find the fibonacci term: "))
print(f"the fibonacci term in {num} :",Fibonacci(num))
```

```python
#function to reverse a string
def revstr(s):
    n=len(s)
    s1=""
    for i in range(n):
        s1=s1+ s[n-1-i]
    return s1
str=input("enter a string to reverse: ")
print(f"the reverse of {str} is:",revstr(str))
```

```python
#function to return the sum of n natural numbers
def sumofnumbers(n):
    sum=0
    for i in range(n+1):
        sum+=i
    return sum
num=int(input("enter a number to find the sum: "))
print(f"the sum of {num} numbers is: ",sumofnumbers(num))
```

```python
#program to find how many even numbers are present in a list
li=input("enter a list of numbers: ")
l=map(int,li.split())
count=0
for i in l :
    if i % 2 == 0:
         count+=1
print(count)
```

```python
#program to print prime numbers in between 1 to 10 
def is_prime(num):
    if num < 2:
        return False
    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            return False
    return True
def print_primes(start, end):
    primes = [n for n in range(start, end + 1) if is_prime(n)]
    print(f"Prime numbers between {start} and {end}:")
    print(primes)
if __name__ == "__main__":
    print_primes(1, 100)
```
